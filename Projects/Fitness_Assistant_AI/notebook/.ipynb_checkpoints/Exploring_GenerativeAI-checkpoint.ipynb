## Objective
The goal of this exercise was to fine-tune an OpenAI GPT-4o-mini model using a small custom dataset (`FitnessCoach-50.jsonl`) containing fitness-related prompts and responses. The intention was to create a specialized AI capable of generating personalized workout plans, mimicking a fitness coach.

This notebook demonstrates the process of fine-tuning an OpenAI GPT model on a small, domain-specific dataset for a fitness coach application. The goal is to generate personalized workout plans with structured outputs including warm-up, main exercises, and cool-down.

The workflow includes:
1. Loading API keys and environment setup
2. Dataset validation
3. Uploading the dataset
4. Creating a fine-tuning job
5. Polling the job status until completion
6. Evaluate the tuned Model
7. Testing the fine-tuned model

## Step 1: Environment Setup & API Key - Load OpenAI API key and prepare environment for requests.

- Load the API key from a `.env` file using `python-dotenv`.  
- Verify that the API key exists to prevent runtime errors.  
- Define constants for the API base URL and headers to simplify API calls.

This step ensures secure and reliable access to the OpenAI API. Using environment variables avoids hardcoding sensitive information and makes the workflow more portable.


```python
import os
import json
import time
import requests
from dotenv import load_dotenv

# Load API key
load_dotenv("ApiKey.env")
API_KEY = os.getenv("OPENAI_API_KEY")
if not API_KEY:
    raise ValueError("OPENAI_API_KEY not found in environment file")
print("OpenAI API key loaded successfully!")

BASE_URL = "https://api.openai.com/v1"
HEADERS = {"Authorization": f"Bearer {API_KEY}"}
```

## Step 2: Dataset Validation - Confirm that the training dataset exists before attempting upload.

- Use `os.path.isfile` to check if the dataset file is present.  
- Ensure the dataset is in JSONL format, as required by OpenAI for fine-tuning.  
- Early validation prevents errors later in the workflow.

This is especially important for small datasets where missing or misformatted files could halt the entire fine-tuning process. For larger datasets, additional sanity checks such as verifying the number of lines or sample content could be useful.


```python
# Path to existing dataset
dataset_path = "../data/FitnessCoach-50.jsonl"
if not os.path.isfile(dataset_path):
    raise FileNotFoundError(f"Dataset file not found: {dataset_path}")
print(f"Dataset file found: {dataset_path}")
```

## Step 3: Upload Dataset - Send the dataset to OpenAI to prepare for fine-tuning.

- Use the `/files` endpoint with `purpose="fine-tune"` to mark the file for training.  
- Capture the returned `file_id` for use in fine-tuning requests.  
- Validate the upload response to ensure the dataset was received successfully.

Uploading small datasets is fast, but large datasets may require splitting or chunking. Always check the response for errors such as invalid formatting or rate limits.


```python
# Upload the dataset
with open(dataset_path, "rb") as f:
    files = {"file": (dataset_path, f)}
    data = {"purpose": "fine-tune"}
    upload_resp = requests.post(f"{BASE_URL}/files", headers=HEADERS, files=files, data=data)

if upload_resp.status_code != 200:
    raise Exception(f"File upload failed: {upload_resp.text}")

file_id = upload_resp.json()["id"]
print("Dataset uploaded successfully. File ID:", file_id)
```

## Step 4: Create Fine-Tuning Job - Request OpenAI to fine-tune a base model using the dataset.

- Choose a base model (e.g., `gpt-4.1-2025-04-14`) for advanced reasoning.  
- Set `n_epochs` and `learning_rate_multiplier` for small datasets to balance learning and avoid overfitting.  
- Add a `suffix` to identify the resulting fine-tuned model.  
- Capture the `job_id` for polling and tracking progress.

Fine-tuning small datasets can overfit quickly. Hyperparameters may need adjustment based on the outputs.


```python
# Create fine-tune job using GPT-4o-mini
payload = {
    "training_file": file_id,
    "model": "gpt-4.1-2025-04-14",
    "hyperparameters": {
        "n_epochs": 3,
        "learning_rate_multiplier": 1.5
    },
    "suffix": "FitnessCoach-50"
}

resp = requests.post(
    f"{BASE_URL}/fine_tuning/jobs",
    headers={**HEADERS, "Content-Type": "application/json"},
    data=json.dumps(payload)
)

if resp.status_code != 200:
    raise Exception(f"Fine-tune job creation failed: {resp.text}")

fine_tune_job = resp.json()
job_id = fine_tune_job["id"]
print("Fine-tune job created successfully. Job ID:", job_id)
```

## Step 5: Poll Fine-Tuning Job Status - Monitor the fine-tuning process until completion.

- Periodically request job status using the `job_id`.  
- Handle network errors gracefully to avoid script crashes.  
- Exit the loop when the job `status` is `succeeded` or `failed`.

Polling every 30 seconds is sufficient for small jobs. Large datasets may take longer, and proper error handling is essential to prevent interruptions.


```python
# Poll fine-tune job status until finished
print("Polling fine-tune job status...")

while True:
    try:
        status_resp = requests.get(f"{BASE_URL}/fine_tuning/jobs/{job_id}", headers=HEADERS)
        if status_resp.status_code != 200:
            print(f"Error fetching job status: {status_resp.text}")
            time.sleep(30)
            continue
        
        job_status = status_resp.json()
        status = job_status.get("status")
        print("Current job status:", status)
        
        if status in ["succeeded", "failed"]:
            break
        time.sleep(30)
    except requests.exceptions.RequestException as e:
        print(f"Network error: {e}. Retrying in 30 seconds...")
        time.sleep(30)
```

### Retrieve Fine-Tuned Model - Obtain the fine-tuned model identifier for API calls.

- If the job succeeded, extract `fine_tuned_model`.  
- Verify that the model name exists; otherwise, investigate the job details.

Retrieving the model ensures we can test and use it. Occasionally, a fine-tuning job may succeed but not return a usable model due to dataset formatting issues.


```python
# Check fine-tuned model
if status == "succeeded":
    fine_tuned_model = job_status.get("fine_tuned_model")
    if fine_tuned_model:
        print("Fine-tuned model ready:", fine_tuned_model)
    else:
        raise ValueError("Fine-tuned model name not found. Check OpenAI dashboard for details.")
else:
    raise RuntimeError(f"Fine-tuning failed. Job details: {job_status}")
```

## Step 6: Evaluate the Fine-Tuned Model

This section evaluates the performance of our **fine-tuned fitness coach AI** for generating workout plans. We test the model using a set of sample prompts and collect metrics such as **token usage, accuracy, and simulated loss**. The goal is to assess how well the fine-tuned model produces structured, beginner-friendly workout routines.

### **Evaluation Process**

1. **Load API and Model**
   - The OpenAI API key is loaded from an environment file.
   - The fine-tuned model ID is specified (`fine_tuned_model`).
   - API headers and endpoints are set up for requests.

2. **Define Test Prompts**
   - We use a set of sample prompts to evaluate the model:
     - Generate a 5-day beginner workout plan for a 30-year-old female.
     - Generate a 3-day full-body home workout plan for a 25-year-old male.
     - Provide a 7-day cardio-focused workout routine for a 40-year-old female.

3. **Query Fine-Tuned Model**
   - Each prompt is sent to the fine-tuned model using a `POST` request.
   - Model responses (`outputs`) and token usage (`token_counts`) are collected.

4. **Compute Approximate Accuracy**
   - We define a set of **expected exercises** (`Squats, Push-ups, Lunges, Plank, Glute Bridge, Jumping Jacks`).
   - For each output, the accuracy is calculated as the fraction of expected exercises present in the response.
   - Simulated loss is computed as `1 - accuracy`.

5. **Display Metrics**
   - Metrics are displayed both as a **console table** and optionally as an **HTML table** in the notebook.
   - Columns include:
     - Prompt
     - Tokens used
     - Accuracy
     - Simulated Loss

6. **Visualize Results**
   - A plot is generated with:
     - **Simulated Loss** on the left y-axis (red line)
     - **Accuracy** on the right y-axis (green dashed line)
     - X-axis shows test prompts
   - This provides a visual understanding of model performance across different scenarios.

7. **Save Summary**
   - All relevant details (prompts, outputs, token counts, accuracy, and simulated loss) are saved into a JSON file:  
     `workout_finetune_summary.json`
   - This allows easy retrieval for reporting or further analysis.


```python
## Summary
# --- Imports ---
import os
import json
import requests
import matplotlib.pyplot as plt
from pathlib import Path
from dotenv import load_dotenv
from IPython.display import display, HTML

# --- Load API key ---
load_dotenv("ApiKey.env")
API_KEY = os.getenv("OPENAI_API_KEY")
if not API_KEY:
    raise ValueError("OPENAI_API_KEY not found in environment file")
print("OpenAI API key loaded successfully!")

# --- Setup API base and headers ---
BASE_URL = "https://api.openai.com/v1"
HEADERS = {"Authorization": f"Bearer {API_KEY}"}

# --- Fine-tuned model info ---
# fine_tuned_model = "ft:gpt-4.1-2025-04-14:personal:fitnesscoach-50:CXacFjXB"

# --- Sample test prompts ---
test_prompts = [
    "Generate a 5-day beginner workout plan for a 30-year-old female.",
    "Generate a 3-day full-body home workout plan for a 25-year-old male.",
    "Provide a 7-day cardio-focused workout routine for a 40-year-old female."
]

outputs = []
token_counts = []

# --- Query fine-tuned model and collect outputs ---
for prompt in test_prompts:
    response = requests.post(
        f"{BASE_URL}/chat/completions",
        headers={**HEADERS, "Content-Type": "application/json"},
        data=json.dumps({
            "model": fine_tuned_model,
            "messages": [
                {"role": "system", "content": "You are a fitness coach AI."},
                {"role": "user", "content": prompt}
            ],
            "temperature": 0.7
        })
    )
    if response.status_code == 200:
        output_text = response.json()["choices"][0]["message"]["content"]
        outputs.append(output_text)
        token_counts.append(response.json()["usage"]["total_tokens"])
    else:
        outputs.append(f"Error: {response.text}")
        token_counts.append(0)

# --- Approximate metrics ---
expected_exercises = ["Squats", "Push-ups", "Lunges", "Plank", "Glute Bridge", "Jumping Jacks"]

accuracy_scores = []
for out in outputs:
    count = sum([1 for ex in expected_exercises if ex.lower() in out.lower()])
    accuracy = count / len(expected_exercises)
    accuracy_scores.append(round(accuracy, 2))

loss_scores = [1 - a for a in accuracy_scores]

# --- Display metrics table ---
table_html = "<table border='1' cellpadding='5'><tr><th>Prompt</th><th>Tokens</th><th>Accuracy</th><th>Simulated Loss</th></tr>"
for i, prompt in enumerate(test_prompts):
    table_html += f"<tr><td>{prompt}</td><td>{token_counts[i]}</td><td>{accuracy_scores[i]}</td><td>{loss_scores[i]}</td></tr>"
table_html += "</table>"

# # Display in notebook
# print("\n--- Fine-Tuned Model Evaluation Metrics ---")
# display(HTML(table_html))

# Print table to console
print("\n--- Fine-Tuned Model Evaluation Metrics ---")
print(f"{'Prompt':<70} {'Tokens':<6} {'Accuracy':<8} {'Simulated Loss':<14}")
for i, prompt in enumerate(test_prompts):
    print(f"{prompt[:65]:<70} {token_counts[i]:<6} {accuracy_scores[i]:<8} {loss_scores[i]:<14}")

# --- Plot metrics ---
fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.set_xlabel("Test Prompts")
ax1.set_ylabel("Simulated Loss", color="tab:red")
ax1.plot(range(len(loss_scores)), loss_scores, color="tab:red", marker="o", label="Simulated Loss")
ax1.tick_params(axis='y', labelcolor="tab:red")
ax1.set_xticks(range(len(test_prompts)))
ax1.set_xticklabels([f"Prompt {i+1}" for i in range(len(test_prompts))], rotation=15)

ax2 = ax1.twinx()
ax2.set_ylabel("Accuracy", color="tab:green")
ax2.plot(range(len(accuracy_scores)), accuracy_scores, color="tab:green", marker="x", linestyle='--', label="Accuracy")
ax2.tick_params(axis='y', labelcolor="tab:green")

lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines + lines2, labels + labels2, loc="upper right")
plt.title("Workout Fine-Tuned Model: Simulated Accuracy & Loss")
plt.grid(True)
plt.show()

# --- Save summary ---
summary = {
    "fine_tuned_model": fine_tuned_model,
    "test_prompts": test_prompts,
    "test_outputs": outputs,
    "token_counts": token_counts,
    "accuracy_scores": accuracy_scores,
    "simulated_loss": loss_scores
}

summary_file = Path("../appendix/workout_finetune_summary.json")
with open(summary_file, "w", encoding="utf-8") as fh:
    json.dump(summary, fh, indent=2, ensure_ascii=False)
print("Summary saved:", summary_file.resolve())
```

## Fine-Tuned Workout Model Evaluation Summary

### **Observations & Insights**

- **Prompt Length & Complexity:**  
  Short, clear prompts achieve **perfect accuracy (1.0)**. More complex prompts, like the 7-day cardio routine, show **much lower accuracy (0.17)**, indicating that the model struggles with multi-faceted or detailed instructions.

- **Token Usage:**  
  Token counts vary (246–398). Even moderate token outputs can show **low accuracy** if the prompt is complex, highlighting that token length alone doesn’t guarantee correctness.

- **Simulated Loss:**  
  Calculated as `1 - Accuracy`. Minimal loss (0.0) for simple prompts confirms reliable output for standard beginner plans. The high loss (0.83) for complex prompts underscores **omissions or incomplete workout details**.

- **Consistency & Structure:**  
  Fine-tuning improves **uniform formatting, bullet points, and headings**, making outputs readable and beginner-friendly, especially for simpler prompts.

- **Practicality & Realism:**  
  Simple plans are realistic and actionable. However, complex plans may lack **exercise variety, progression guidance, or full coverage**, due to limited training samples.

- **Impact of Fine-Tuning:**  
  Fine-tuning clearly enhances **structured output and alignment with project-specific prompts**. Standard beginner routines are handled well, while more detailed or long prompts need refinement.

- **Future Improvements:**  
  * Expand dataset size and diversity to improve handling of complex scenarios.  
  * Include **progressive difficulty, alternative exercises, and active recovery options**.  
  * Add realistic **rep ranges, session timing, and equipment variations** for improved usability.

**Insight:**  
Fine-tuning has successfully optimized the model for **clear, structured, beginner-focused workout plans**, but additional dataset diversity and complex prompt handling are required to ensure accuracy in longer or multi-faceted routines.

- Provide a structured prompt requesting a 25-minute workout.  
- Use system messages to set context: AI acts as a fitness coach.  
- Set `temperature=0.7` to balance creativity with adherence to instructions.

Testing ensures the fine-tuning produced domain-specific improvements. With small datasets, outputs may be repetitive or simplistic; iterative improvement with more examples or hyperparameter tuning may be necessary.


```python
# Test fine-tuned model
test_prompt = "Generate a 5-day beginner workout plan for a 30-year-old female."

response = requests.post(
    f"{BASE_URL}/chat/completions",
    headers={**HEADERS, "Content-Type": "application/json"},
    data=json.dumps({
        "model": fine_tuned_model,
        "messages": [
            {"role": "system", "content": "You are a fitness coach AI."},
            {"role": "user", "content": test_prompt}
        ],
        "temperature": 0.7
    })
)

if response.status_code == 200:
    output = response.json()["choices"][0]["message"]["content"]
    print("=== Test Prompt ===")
    print(test_prompt)
    print("=== Model Output ===")
    print(output)
else:
    raise Exception(f"Error calling fine-tuned model: {response.text}")
```

   * **Fine-Tuned 4.1:** Exercises **repeat across multiple days** (Squats, Push-ups, Bent Over Rows, Glute Bridge), with little variation.
   * **Base Model:** Offers **greater variety** across days, including strength, cardio, flexibility, and active recovery. Also includes optional exercises (e.g., dumbbells, interval walks).

3. **Beginner Guidance & Practicality**

   * **Fine-Tuned 4.1:** Provides basic structure but minimal additional guidance; no tips for rest intervals or progression.
   * **Base Model:** Provides detailed instructions, safety tips, hydration advice, and progression guidance, making it more beginner-friendly.

4. **Realism & Usability**

   * **Fine-Tuned 4.1:** Highly structured, but somewhat repetitive; lacks personalization and progression.

5. **Overall Impression**

   * The base model still provides **more variety, practical advice, and realism**, making it closer to a human-designed beginner workout plan.

**Conclusion:**

* Fine-tuning GPT-4.1 produced **clear, readable, and structured workout plans** suitable for beginner guidance.
* However, **exercise diversity and progression guidance** are areas for improvement.
* Dataset quality and variety remain crucial in achieving realistic and personalized AI-generated plans.

## Fine-Tuning with 3.5 (Additional)

The initial fine-tuning was done for GPT-3.5 with 500 training samples (FitnessCoach-500.jsonl) and successfully generated a functional model. However, upon re-running with few modifications, the model was **blocked due to policy violations**, flagged by OpenAI’s moderation system for producing potentially unsafe outputs. OpenAI recommends using newer base models like **GPT-4.1**, which are safer, more capable, and cost-effective, which is presented above. However, the outputs are presented here for **comparison with the base model**.

```python
# Test the Fine-Tuned GPT-3.5-Turbo Model
import os
import json
import requests
from dotenv import load_dotenv

# Load API key
load_dotenv("ApiKey.env")
API_KEY = os.getenv("OPENAI_API_KEY")
if not API_KEY:
    raise ValueError("OPENAI_API_KEY not found in environment file")

BASE_URL = "https://api.openai.com/v1"
HEADERS = {"Authorization": f"Bearer {API_KEY}"}

fine_tuned_model = "ft:gpt-3.5-turbo-0125:personal::CXXFEioT"

# Define test prompts
test_prompts = [
    "Generate a 5-day beginner workout plan for a 30-year-old female.",
    "Create a 3-day strength training plan for a 40-year-old male.",
    "Suggest a low-impact workout for a 50-year-old beginner.",
]

# Call the fine-tuned model for each prompt
for prompt in test_prompts:
    response = requests.post(
        f"{BASE_URL}/chat/completions",
        headers={**HEADERS, "Content-Type": "application/json"},
        data=json.dumps({
            "model": fine_tuned_model,
            "messages": [
                {"role": "system", "content": "You are a fitness coach AI."},
                {"role": "user", "content": prompt}
            ],
            "temperature": 0.7
        })
    )
    
    if response.status_code == 200:
        output = response.json()["choices"][0]["message"]["content"]      
        print("=== Test Prompt ===")
        print(prompt)
        print("=== Model Output ===")
        print(output)
        print("\n" + "-"*50 + "\n")
    else:
        print(f"Error calling fine-tuned model: {response.text}")
```

4. **Practicality & Safety**

* For GPT-3.5, fine-tuning on a **smaller dataset (500 samples)** produced structured but repetitive outputs, with limited exercise variety and minimal guidance.
* For GPT-4.1, fine-tuning on a **more diverse and focused dataset** resulted in clearer formatting, consistent structure, and beginner-friendly readability. However, exercise variety and progression guidance still did not fully match the base model’s richness.

**Conclusion:**

* Increasing dataset quality and size improved **consistency and readability** of fine-tuned outputs.
* Fine-tuning made the model **more domain-specific**, but did not always surpass the base model in exercise diversity or practical guidance, highlighting the importance of dataset diversity and careful prompt design.
