```python
# ============================================
# Airline Delay Prediction - Sampled from Complete Rows
# ============================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.impute import SimpleImputer

import warnings
warnings.filterwarnings("ignore")

# --------------------------------------------
# Paths
# --------------------------------------------
DATA_FOLDER = "../data/"
FIG_PATH = "../figures/"
MODEL_PATH = "../code/"

os.makedirs(FIG_PATH, exist_ok=True)
os.makedirs(MODEL_PATH, exist_ok=True)

# --------------------------------------------
# Files and columns to use
# --------------------------------------------
files = [f for f in os.listdir(DATA_FOLDER) if f.endswith(".csv")]
selected_years = ["2018","2019","2020"]
cols_needed = ["FL_DATE","OP_CARRIER","OP_CARRIER_FL_NUM","ORIGIN","DEST",
               "CRS_DEP_TIME","DEP_TIME","DEP_DELAY","TAXI_OUT","WHEELS_OFF","WHEELS_ON",
               "TAXI_IN","CRS_ARR_TIME","ARR_TIME","ARR_DELAY","CANCELLED",
               "DIVERTED","CRS_ELAPSED_TIME","ACTUAL_ELAPSED_TIME","DISTANCE"]

# --------------------------------------------
# Sample rows from each year's dataset
# --------------------------------------------
sampled_list = []

for file in files:
    if any(year in file for year in selected_years):
        df_year = pd.read_csv(os.path.join(DATA_FOLDER, file))
        df_year = df_year[[col for col in cols_needed if col in df_year.columns]]
        df_year_complete = df_year.dropna()
        if df_year_complete.shape[0] > 0:
            sample_size = min(1000, df_year_complete.shape[0])
            df_sampled = df_year_complete.sample(n=sample_size, random_state=42)
            sampled_list.append(df_sampled)

df = pd.concat(sampled_list, ignore_index=True)
print("Combined dataset shape (sampled 1000 rows per year):", df.shape)

df = df.drop(columns=["CANCELLATION_CODE","Unnamed: 27"], errors="ignore")
df = df[(df["CANCELLED"] == 0) & (df["DIVERTED"] == 0)]
print("After filtering cancelled/diverted flights:", df.shape)

# --------------------------------------------
# Target variables
# --------------------------------------------
df["DELAYED"] = (df["ARR_DELAY"] > 15).astype(int)
y_class = df["DELAYED"]
y_reg = df["ARR_DELAY"]

# --------------------------------------------
# Feature engineering
# --------------------------------------------
df["FL_DATE"] = pd.to_datetime(df["FL_DATE"])
df["MONTH"] = df["FL_DATE"].dt.month
df["DAY_OF_WEEK"] = df["FL_DATE"].dt.dayofweek
df["DEP_HOUR"] = df["CRS_DEP_TIME"] // 100
df["TIME_BLOCK"] = pd.cut(df["DEP_HOUR"], bins=[0,6,12,18,24],
                          labels=["Night","Morning","Afternoon","Evening"], include_lowest=True)

features = ["OP_CARRIER", "ORIGIN", "DEST", "DISTANCE", "CRS_ELAPSED_TIME", "MONTH", "DAY_OF_WEEK", "TIME_BLOCK"]
X = df[features]
categorical_cols = ["OP_CARRIER", "ORIGIN", "DEST", "TIME_BLOCK"]
numeric_cols = ["DISTANCE", "CRS_ELAPSED_TIME", "MONTH", "DAY_OF_WEEK"]

categorical_transformer = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                                    ("onehot", OneHotEncoder(handle_unknown="ignore"))])
numeric_transformer = Pipeline([("imputer", SimpleImputer(strategy="median"))])
preprocessor = ColumnTransformer([("num", numeric_transformer, numeric_cols),
                                  ("cat", categorical_transformer, categorical_cols)])

# ============================================
# Classification Model
# ============================================
X_train, X_test, y_train, y_test = train_test_split(X, y_class, test_size=0.2, random_state=42)
clf = Pipeline([("preprocessor", preprocessor),
                ("model", RandomForestClassifier(n_estimators=50, random_state=42))])
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
y_prob = clf.predict_proba(X_test)[:,1]

# ============================================
# Regression Model
# ============================================
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X, y_reg, test_size=0.2, random_state=42)
reg = Pipeline([("preprocessor", preprocessor),
                ("model", RandomForestRegressor(n_estimators=50, random_state=42))])
reg.fit(X_train_r, y_train_r)
y_pred_r = reg.predict(X_test_r)

# ============================================
# Results Table Display
# ============================================
clf_results = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "ROC-AUC"],
    "Value": [accuracy_score(y_test, y_pred),
              precision_score(y_test, y_pred),
              recall_score(y_test, y_pred),
              roc_auc_score(y_test, y_prob)]
})

reg_results = pd.DataFrame({
    "Metric": ["MAE", "RMSE"],
    "Value": [mean_absolute_error(y_test_r, y_pred_r),
              np.sqrt(mean_squared_error(y_test_r, y_pred_r))]
})

print("\n--- Classification Results ---")
print(clf_results.to_string(index=False))

print("\n--- Regression Results ---")
print(reg_results.to_string(index=False))

# Save models
joblib.dump(clf, MODEL_PATH + "delay_classification_model_sample.pkl")
joblib.dump(reg, MODEL_PATH + "delay_regression_model_sample.pkl")

# ============================================
# Feature Importance
# ============================================
model = clf.named_steps["model"]
feature_names = clf.named_steps["preprocessor"].get_feature_names_out()
importances = model.feature_importances_
importance_df = pd.DataFrame({"Feature": feature_names, "Importance": importances}).sort_values(by="Importance", ascending=False)
importance_df.head(15).to_csv(MODEL_PATH + "feature_importance_sample.csv", index=False)

plt.figure(figsize=(10,6))
sns.barplot(x="Importance", y="Feature", data=importance_df.head(15))
plt.title("Top 15 Feature Importances (Classification)")
plt.tight_layout()
plt.savefig(FIG_PATH + "feature_importance_sample.png")
plt.show()
plt.close()

# ============================================
# Visualizations
# ============================================
plt.figure()
sns.histplot(df["ARR_DELAY"], bins=50, kde=True)
plt.title("Distribution of Arrival Delays (Sampled)")
plt.savefig(FIG_PATH + "delay_distribution_sample.png")
plt.show()
plt.close()

plt.figure(figsize=(10,6))
sns.boxplot(x="OP_CARRIER", y="ARR_DELAY", data=df)
plt.title("Arrival Delay by Carrier (Sampled)")
plt.savefig(FIG_PATH + "delay_by_carrier_box_sample.png")
plt.show()
plt.close()

plt.figure()
sns.scatterplot(x="DISTANCE", y="ARR_DELAY", data=df)
plt.title("Distance vs Arrival Delay")
plt.savefig(FIG_PATH + "distance_vs_delay_sample.png")
plt.show()
plt.close()

plt.figure()
sns.barplot(x="DAY_OF_WEEK", y="ARR_DELAY", data=df)
plt.title("Average Arrival Delay by Day of Week")
plt.savefig(FIG_PATH + "delay_by_day_sample.png")
plt.show()
plt.close()

plt.figure()
sns.barplot(x="TIME_BLOCK", y="ARR_DELAY", data=df)
plt.title("Average Arrival Delay by Time Block")
plt.savefig(FIG_PATH + "delay_by_timeblock_sample.png")
plt.show()
plt.close()

plt.figure()
sns.scatterplot(x="CRS_ELAPSED_TIME", y="ARR_DELAY", data=df)
plt.title("Scheduled Elapsed Time vs Arrival Delay")
plt.savefig(FIG_PATH + "crs_elapsed_vs_delay_sample.png")
plt.show()
plt.close()
```
